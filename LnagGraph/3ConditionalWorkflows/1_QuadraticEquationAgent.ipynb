```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class QuadraticState(TypedDict):
    a: int
    b: int
    c: int

    equation: str
    discriminant: float
    result: str
```

```python
graph = StateGraph(QuadraticState)
```

```python
def get_equation(state: QuadraticState):
    #given a, b, c, return the equation as a string
    equation = f"{state['a']}x^2 + ({state['b']}x) + ({state['c']}) = 0"
    return {"equation": equation}
```

```python
def calculate_discriminant(state: QuadraticState):
    a = state['a']
    b = state['b']
    c = state['c']
    discriminant = b**2 - 4*a*c
    
    return {"discriminant": discriminant}
```

```python
def real_roots(state: QuadraticState):
    a = state['a']
    b = state['b']
    c = state['c']
    D = state['discriminant']
    root1 = (-b + D**0.5) / (2*a)
    root2 = (-b - D**0.5) / (2*a)
    result = {"result": f"The Quadratic equation has two real roots: {root1}, {root2}"}
    return {'result': result}


def complex_roots(state: QuadraticState):
    result = "The Quadratic equation has no real roots."
    return {'result': result}

def equal_roots(state: QuadraticState):
    a = state['a']
    b = state['b']
    root = -b / (2*a)
    result = {"result": f"The Quadratic equation has one real root: {root}"}
    return {'result': result}
```

```python
def conditial_discriminant(state: QuadraticState) -> Literal["real_roots", "equal_roots", "complex_roots"]  :
    D = state['discriminant']
    if D > 0:
        return "real_roots"
    elif D == 0:
        return "equal_roots"
    else:
        return "complex_roots"
```

```python
#add nodes to the graph
graph.add_node("get_equation", get_equation)
graph.add_node("calculate_discriminant", calculate_discriminant)
graph.add_node("real_roots", real_roots)
graph.add_node("complex_roots", complex_roots)
graph.add_node("equal_roots", equal_roots)

#add edges to the graph
graph.add_edge(START, "get_equation")
graph.add_edge("get_equation", "calculate_discriminant")
graph.add_conditional_edges("calculate_discriminant", conditial_discriminant)
graph.add_edge("real_roots", END)
graph.add_edge("complex_roots", END)
graph.add_edge("equal_roots", END)
```

```python
workflow = graph.compile()
```

```python
workflow.get_graph().print_ascii()
```
